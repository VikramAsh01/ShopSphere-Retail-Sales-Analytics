```python
import pandas as pd

df = pd.read_csv("C:/Users/lenovo/Downloads/ShopSphere-Customer-Intelligence/data/cleaned/superstore_cleaned.csv")

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
customers = df[
    [
        "Customer ID",
        "Customer Name",
        "Segment"
    ]
].drop_duplicates()

customers.head()
```

```python
customers.to_csv(
    "C:/Users/lenovo/Downloads/ShopSphere-Customer-Intelligence/data/cleaned/customers.csv",
    index=False
)
```

```python
products = (
    df[
        [
            "Product ID",
            "Product Name",
            "Category",
            "Sub-Category"
        ]
    ]
    .drop_duplicates(subset=["Product ID"])
)

products.head()
products.to_csv(
    "C:/Users/lenovo/Downloads/ShopSphere-Customer-Intelligence/data/cleaned/products.csv",
    index=False
)
```

```python
orders = df[
    [
        "Order ID",
        "Order Date",
        "Ship Date",
        "Ship Mode",
        "Customer ID",
        "Country",
        "City",
        "State",
        "Postal Code",
        "Region"
    ]
].drop_duplicates()

orders.head()
orders.to_csv(
   "C:/Users/lenovo/Downloads/ShopSphere-Customer-Intelligence/data/cleaned/orders.csv",
       index=False
)
```

```python
sales = df[
    [
        "Row ID",
        "Order ID",
        "Product ID",
        "Sales",
        "Quantity",
        "Discount",
        "Profit"
    ]
]

sales.head()
sales.to_csv(
    "C:/Users/lenovo/Downloads/ShopSphere-Customer-Intelligence/data/cleaned/sales.csv",
    index=False
)
```
