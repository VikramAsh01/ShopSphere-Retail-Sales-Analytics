```python
clean_df = df.copy()
```

```python
clean_df["Order Date"] = pd.to_datetime(clean_df["Order Date"])

clean_df["Ship Date"] = pd.to_datetime(clean_df["Ship Date"])
clean_df.info()
```

```python
clean_df[clean_df["Sales"] < 0]
clean_df[clean_df["Profit"] < 0]
clean_df["Discount"].describe()
```

```python
text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()
```

```python
clean_df.duplicated().sum()
```

```python
clean_df["Delivery Days"] = (
    clean_df["Ship Date"] - clean_df["Order Date"]
).dt.days
```

```python
clean_df["Order Year"] = clean_df["Order Date"].dt.year
```

```python
clean_df["Order Month"] = clean_df["Order Date"].dt.month_name()
```

```python
clean_df["Month Number"] = clean_df["Order Date"].dt.month
```

```python
clean_df["Quarter"] = clean_df["Order Date"].dt.quarter
```

```python
clean_df["Weekday"] = clean_df["Order Date"].dt.day_name()
```

```python
clean_df["Weekend Order"] = clean_df["Weekday"].map(
    {True: "Weekend", False: "Weekday"}
)
```

```python
clean_df["Profit Margin %"] = (
    clean_df["Profit"] / clean_df["Sales"]
) * 100
```

```python
clean_df["Sales Bucket"] = pd.cut(
    clean_df["Sales"],
    bins=[0,100,500,1000,5000,25000],
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
clean_df["Discount Category"] = pd.cut(
    clean_df["Discount"],
    bins=[-0.01,0,0.2,0.5,1],
    labels=[
        "No Discount",
        "Low",
        "Medium",
        "High"
    ]
)
```
